```python
from glob import glob
from pathlib import Path
import json
import random
from collections import defaultdict
```

```python
fn_test = "../saves/reproduce/HGERE/gsap_rel_models/scibert/tersibcop/facencbiaf-seq512-mem400-iter3-layernorm+_attnself/ent400-rel400-lr2e-5-1e-4-bs18-ep10-eps1e-8/Hyper_scierc_scibert-42/pred_ent_pred_test.json"
```

```python
docs = []
with open(fn_test) as f:
    for line in f.readlines():
        doc = json.loads(line)
        docs.append(doc)
```

```python
doc = random.choice(docs)
```

```python
doc.keys()
```

```python
sent_idxs = list(range(len(doc["sentences"])))
print(len(sent_idxs))
sent_idxs_selected = []
for idx in sent_idxs:
    if doc["ner"][idx] or doc["predicted_ner"][idx]:
        sent_idxs_selected.append(idx)
print(len(sent_idxs_selected))
```

```python
## choose random sentence
sent_idx = random.choice(sent_idxs_selected)
print(f"sentence_idx:", sent_idx)## prepare sent (incl. ents and relations)
sent = " ".join(doc["sentences"][sent_idx])
ent_dict = defaultdict(list)
for start, end, label in doc["ner"][sent_idx]:
    ent_token = []
    for token_idx in range(start, end + 1):
        token = doc["doc_tokens"][token_idx]
        ent_token.append(token)
    ent_text = " ".join(ent_token)
    ent_print = f"[{label}: '{ent_text}']"
    ent_dict[(start, end)].append(ent_print)
rels_print = []
for subj_start, subj_end, obj_start, obj_end, label in doc["relations"][sent_idx]:
    subj = ", ".join(ent_dict[(subj_start, subj_end)])
    obj = ", ".join(ent_dict[(obj_start, obj_end)])
    rel_print = f"{subj} - {label} -> {obj}"
    rels_print.append(rel_print)

pred_ent_dict = defaultdict(list)
for start, end, label in doc["predicted_ner"][sent_idx]:
    ent_token = []
    for token_idx in range(start, end + 1):
        token = doc["doc_tokens"][token_idx]
        ent_token.append(token)
    ent_text = " ".join(ent_token)
    ent_print = f"[{label}: '{ent_text}']"
    pred_ent_dict[(start, end)].append(ent_print)
pred_rels_print = []
for subj_start, subj_end, obj_start, obj_end, label in doc["predicted_rel"][sent_idx]:
    subj = ", ".join(pred_ent_dict[(subj_start, subj_end)])
    obj = ", ".join(pred_ent_dict[(obj_start, obj_end)])
    rel_print = f"{subj} - {label} -> {obj}"
    pred_rels_print.append(rel_print)

## print sentence
print(sent)
print("NER")
for ent_print in ent_dict.values():
    print("", " ".join(ent_print))
print("RELATIONS")
for rel in rels_print:
    print("", rel)
print()
print("NER predicted")
for ent_print in pred_ent_dict.values():
    print("", " ".join(ent_print))
print("RELATIONS predicted")
for rel in pred_rels_print:
    print("", rel)
```
